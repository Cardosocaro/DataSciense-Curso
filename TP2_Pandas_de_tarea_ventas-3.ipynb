## Parte 3 — Diagnóstico de la columna `fecha`

La columna `fecha` debe poder ordenarse y agruparse por períodos. Antes de convertirla, investigá cómo está escrita.

1. Mostrá valores de fecha y sus frecuencias.
2. Verificá si todos tienen el mismo formato. Usá coincidencias de patrones para distinguir fechas con año al comienzo de fechas con año al final.
3. Contá cuántas filas corresponden a cada patrón y confirmá si entre ambos explican todos los registros.
4. Explicá por qué una fecha como `03-01-2026` puede ser ambigua si no conocemos el formato de origen.

**Concepto:** una fecha como texto no es todavía una fecha para pandas. Convertirla sin conocer la convención puede cambiar silenciosamente el significado de día y mes.

```python
df["fecha"].value_counts()
```

```python
patron_ymd = df["fecha"].str.match(r"^\d{4}-\d{2}-\d{2}$", na=False)

patron_ymd
```

```python
patron_dmy = df["fecha"].str.match(r"^\d{2}-\d{2}-\d{4}$", na=False)

patron_dmy
```

```python
patron_ymd.sum()
```

```python
patron_dmy.sum()
```

```python
patron_ymd.sum() + patron_dmy.sum()
```

Hay fechas en dos tipos de formato  (yyyy-mm-dd y dd-mm-yyyy).
La fecha 03-01-2026 puede ser ambigua porque, dependiendo del formato utilizado, puede representar el 3 de enero de 2026 si se interpreta como día-mes-año, o el 1 de marzo de 2026 si se interpreta como mes-día-año. Por eso es necesario conocer el formato de origen antes de convertir las fechas.

## Parte 4 — Reparación guiada de fechas

Implementá la reparación sin sobrescribir `df["fecha"]` al comienzo. Trabajá con tres Series temporales:

1. Creá una primera Serie que interprete sólo las fechas con formato año-mes-día. Usá `to_datetime`, declarando el formato y usando `errors="coerce"`.
2. Creá una segunda Serie que interprete sólo las fechas con formato día-mes-año, también con formato explícito y `errors="coerce"`.
3. Revisá los valores faltantes de ambas Series. ¿Por qué cada una deja una parte de las filas como `NaT`?
4. Creá una tercera Serie que combine las dos anteriores mediante `combine_first`.
5. Verificá que la Serie combinada no dejó fechas sin interpretar.
6. Recién entonces creá `df_limpio` como copia de `df` y reemplazá en esa copia la columna `fecha` por la Serie combinada.
7. Confirmá que el tipo final de `fecha` sea un tipo de fecha y hora.

**Justificación:** separar los formatos permite saber exactamente qué regla se aplicó a cada grupo de filas. `errors="coerce"` convierte las conversiones fallidas en `NaT`, lo que facilita detectarlas en vez de ocultarlas.

```python
fecha_ymd = pd.to_datetime(
    df["fecha"].where(patron_ymd),
    format="%Y-%m-%d",
    errors="coerce"
)
```

```python
fecha_dmy = pd.to_datetime(
    df["fecha"].where(patron_dmy),
    format="%d-%m-%Y",
    errors="coerce"
)
```

```python
fecha_ymd.isna().sum()
```

```python
fecha_dmy.isna().sum()
```

Cada Serie deja una parte de los registros como NaT porque cada una fue creada para interpretar únicamente uno de los formatos presentes en la columna fecha. Las fechas que no coinciden con el formato especificado no pueden convertirse y, debido a errors="coerce", se representan como NaT. Esto permite separar los dos formatos sin generar errores.

```python
fecha_combinada = fecha_ymd.combine_first(fecha_dmy)
```

```python
fecha_combinada.isna().sum()
```

```python
df_limpio = df.copy()
```

```python
df_limpio["fecha"] = fecha_combinada
```

```python
df_limpio["fecha"].dtype
```

## Parte 5 — Tratamiento de `total`

1. En registros donde `total` esté presente, verificá si coincide con la multiplicación de `cantidad` por `precio_unitario`.
2. Contá los casos en los que esa relación no se cumple.
3. Si la relación es consistente, completá únicamente los valores faltantes de `total` usando esa regla. Usá el método `fillna` sobre `df_limpio`.
4. Volvé a contar los faltantes y verificá otra vez la relación entre las tres columnas.

**Justificación:** un importe puede completarse sólo si existe una regla de negocio comprobable. No uses la media o la mediana de ventas para imputar este campo: modificarían la facturación y no respetarían la operación real. Si la regla no fuera consistente, el valor debería conservarse como faltante o consultarse con la fuente.

```python
total_calculado = df_limpio["cantidad"] * df_limpio["precio_unitario"]
```

```python
df_limpio["total"].eq(total_calculado)
```

```python
(df_limpio["total"].notna() & (df_limpio["total"] != total_calculado)).sum()
```

```python
df_limpio[
    df_limpio["total"].notna() &
    (df_limpio["total"] != total_calculado)
]
```

```python
df_limpio[df_limpio["total"].isna()]
```

```python
df_limpio["total"] = df_limpio["total"].fillna(total_calculado)
```

```python
df_limpio["total"].isna().sum()
```

```python
(df_limpio["total"] == df_limpio["cantidad"] * df_limpio["precio_unitario"]).all()
```

Se verificó la consistencia de la columna total comparándola con el resultado de multiplicar cantidad por precio_unitario. Los valores de total que estaban completos se conservaron y los valores faltantes se pudieron reconstruir utilizando esta relación. Finalmente, se comprobó que todos los registros cumplen la condición de que total es igual a cantidad por precio_unitario.

## Parte 6 — Normalización de texto y variables útiles

1. Elegí una estrategia uniforme para `nombre` y `apellido`: remové espacios de los extremos y normalizá mayúsculas/minúsculas. Aplicala en `df_limpio`.
2. Mostrá nuevamente frecuencias de esos campos para comprobar el efecto.
3. A partir de `fecha`, creá una columna `anio` y una columna `mes` adecuada para agrupar cronológicamente.
4. Explicá por qué una columna de mes como período año-mes suele ser más útil que sólo el número del mes cuando hay más de un año.

**Precaución:** normalizar texto mejora la consistencia de filtros y conteos, pero no crea un identificador de cliente. Dos registros con el mismo nombre y apellido podrían ser personas distintas.

```python
df_limpio["nombre"] = df_limpio["nombre"].str.strip().str.lower()
```

```python
df_limpio["apellido"] = df_limpio["apellido"].str.strip().str.lower()
```

```python
df_limpio["nombre"].value_counts()
```

```python
df_limpio["apellido"].value_counts()
```

```python
df_limpio["anio"] = df_limpio["fecha"].dt.year
```

```python
df_limpio["fecha"].dt.month
```

```python
df_limpio["mes"] = df_limpio["fecha"].dt.to_period("M")
```

```python
df_limpio[["fecha", "anio", "mes"]].head()
```

Utilizar un período año-mes es más útil que utilizar solamente el número del mes porque permite diferenciar un mismo mes perteneciente a distintos años. Por ejemplo, enero de 2025 y enero de 2026 no se mezclarían al realizar agrupaciones. De esta manera, se pueden analizar correctamente las ventas y la facturación a lo largo del tiempo.
La normalización de nombres y apellidos mejora la consistencia de los datos, pero no permite identificar de manera única a un cliente, ya que dos personas diferentes podrían tener el mismo nombre y apellido.
